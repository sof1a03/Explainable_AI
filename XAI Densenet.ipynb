```python
import os
import cv2
import numpy as np
import gc  
from tqdm import tqdm  

# Define source and destination folders
source_folder = "train"
destination_folder = "updated_train"

# Create the new folder if it doesn’t exist
os.makedirs(destination_folder, exist_ok=True)

# Parameters
batch_size = 5000  # Adjust based on memory availability
image_paths = []

# Collect all frontal image paths first
for root, _, files in os.walk(source_folder):
    for file in files:
        if "_frontal" in file.lower():  # Process only frontal images
            file_path = os.path.join(root, file)
            image_paths.append(file_path)

# Fix total batch count
total_batches = (len(image_paths) + batch_size - 1) // batch_size

# Process images in batches
for batch_idx in range(total_batches):
    batch_start = batch_idx * batch_size
    batch_end = min(batch_start + batch_size, len(image_paths))
    batch = image_paths[batch_start:batch_end]

    print(f"Processing batch {batch_idx + 1} of {total_batches} ({batch_end - batch_start} images)...")

    for file_path in tqdm(batch, desc=f"Batch {batch_idx + 1}/{total_batches}"):
        try:
            # Read image
            img = cv2.imread(file_path)

            if img is None:
                continue  

            # Resize to 224x224
            img_resized = cv2.resize(img, (224, 224))

            # Normalize pixel values to [0,1]
            img_resized = img_resized.astype(np.float32) / 255.0

            # Reconstruct the new save path while keeping the folder structure
            relative_path = os.path.relpath(file_path, source_folder)
            save_path = os.path.join(destination_folder, relative_path)

            # Create directories if they don’t exist
            os.makedirs(os.path.dirname(save_path), exist_ok=True)

            # Save the processed image
            cv2.imwrite(save_path, (img_resized * 255).astype(np.uint8))  # Convert back to [0, 255] for saving

        except Exception as e:
            continue  

    # Manually clear memory after processing each batch
    gc.collect()

print("All frontal images have been resized and normalized.")
```

```python
import os
import pandas as pd
import tensorflow as tf
import cv2
import numpy as np
from tensorflow.keras.applications import DenseNet169
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from sklearn.utils import class_weight

csv_path = "train.csv"  #  CSV file
image_base_path = "updated_train/"  #  preprocessed images

df = pd.read_csv(csv_path)
df = df[df["Frontal/Lateral"] == "Frontal"]
df = df[df["Pleural Effusion"].notna()]  # Remove NaN values
df = df[df["Pleural Effusion"] != -1.0]  # Remove uncertain (-1.0) labels

df["Path"] = df["Path"].apply(lambda x: os.path.join(image_base_path, "/".join(x.split("/")[-3:])))
image_paths = df["Path"].values
labels = df["Pleural Effusion"].astype(np.int32).values  # Ensure labels are int32

y
weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(labels),
    y=labels
)
class_weights = dict(zip(np.unique(labels), weights))
print("Computed class weights:", class_weights)
# If your labels are {0, 1}, this should look like {0: some_float, 1: some_float}.

missing_count = 0

def load_image_and_label(image_path, label):
    global missing_count
    image_path = image_path.numpy().decode()
    image = cv2.imread(image_path)

    if image is None:
        missing_count += 1
        if missing_count % 100 == 0:
            print(f"Skipping {missing_count} missing images...")
        image = np.zeros((224, 224, 3), dtype=np.float32)  # Placeholder black image

    image = cv2.resize(image, (224, 224))
    image = image.astype(np.float32) / 255.0
    return image, label

def parse_function(image_path, label):
    image, label = tf.py_function(load_image_and_label, [image_path, label], [tf.float32, tf.int32])
    image.set_shape((224, 224, 3))
    label.set_shape(())
    return image, label

dataset = tf.data.Dataset.from_tensor_slices((image_paths, labels))
dataset = dataset.map(parse_function, num_parallel_calls=tf.data.AUTOTUNE)

train_size = int(0.8 * len(df))
shuffle_buffer_size = min(len(df), 5000)
batch_size = 8  # Adjust based on GPU memory

train_dataset = (
    dataset
    .shuffle(buffer_size=shuffle_buffer_size)
    .take(train_size)
    .batch(batch_size)
    .prefetch(tf.data.AUTOTUNE)
)

val_dataset = (
    dataset
    .skip(train_size)
    .batch(batch_size)
    .prefetch(tf.data.AUTOTUNE)
)

base_model = DenseNet169(weights="imagenet", include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation="relu")(x)
x = Dense(128, activation="relu")(x)
x = Dense(1, activation="sigmoid")(x)

model = Model(inputs=base_model.input, outputs=x)
model.compile(optimizer=Adam(learning_rate=0.0001), loss="binary_crossentropy", metrics=["accuracy"])

steps_per_epoch = tf.data.experimental.cardinality(train_dataset).numpy()
validation_steps = tf.data.experimental.cardinality(val_dataset).numpy()


history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=10,
    steps_per_epoch=steps_per_epoch,
    validation_steps=validation_steps,
    class_weight=class_weights  # <----- Add this line
)

model.save("weighteddensenet169_pleural_effusion.h5")
print("Model training complete & saved as 'densenet169_pleural_effusion.h5'.")
```
